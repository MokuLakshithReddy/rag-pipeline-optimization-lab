# RAG Pipeline Optimization: Engineering Benchmarks & Trade-Off Analysis

This notebook demonstrates systematic retrieval experiments, comparing:
- Fixed vs. Recursive Chunking
- Dense vs. Sparse (BM25) vs. Hybrid (Reciprocal Rank Fusion)
- Cross-Encoder Reranking quality vs. latency trade-offs
- Evaluation across MRR, NDCG@K, Precision@K, Recall@K, and Latency

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

csv_path = "../reports/benchmark_results.csv"
if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    print("Loaded benchmark results:")
    display(df[['experiment_id', 'retrieval_strategy', 'chunking_strategy', 'mrr', 'ndcg_at_k', 'latency_mean_ms']])
else:
    print("Run 'python src/experiments/runner.py' to generate benchmark results first.")
```

```python
# Plot Retrieval Quality vs Latency
if 'df' in locals() and not df.empty:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # MRR / NDCG Bar Chart
    df.set_index('experiment_id')[['mrr', 'ndcg_at_k']].plot(kind='bar', ax=ax1, colormap='viridis')
    ax1.set_title('Ranking Quality (MRR & NDCG@K)')
    ax1.set_ylabel('Score')
    ax1.set_ylim(0, 1.1)
    
    # Latency Bar Chart
    df.set_index('experiment_id')['latency_mean_ms'].plot(kind='bar', ax=ax2, color='crimson')
    ax2.set_title('Mean Retrieval Latency (ms)')
    ax2.set_ylabel('Milliseconds')
    
    plt.tight_layout()
    plt.show()
```
